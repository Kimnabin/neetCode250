# Products of array except self

## Đề bài
Cho một mảng số nguyên `nums` có độ dài `n`.  
Trả về một mảng `answer` sao cho:
- answer[i] = tích của tất cả các phần tử trong nums ngoại trừ nums[i]

### Yêu cầu
- Không được sử dụng phép chia (division).
- Độ phức tạp thời gian: O(n).
- Độ phức tạp bộ nhớ: O(1) (ngoại trừ output).

---

## Hướng giải quyết

### Ý tưởng 1: Brute Force
- Với mỗi `i`, duyệt toàn bộ mảng và nhân các phần tử trừ `nums[i]`.
- Độ phức tạp: `O(n^2)`.

### Ý tưởng 2: Sử dụng mảng phụ `prefix` và `suffix`
- `prefix[i]` = tích từ đầu đến phần tử trước `i`.
- `suffix[i]` = tích từ cuối đến phần tử sau `i`.
- `answer[i] = prefix[i] * suffix[i]`.
- Độ phức tạp: `O(n)` nhưng dùng thêm `O(n)` bộ nhớ.

### Ý tưởng 3 (Tối ưu): Dùng output array như prefix + biến phụ cho suffix
- Bước 1: duyệt trái → phải, lưu prefix vào `answer`.
- Bước 2: duyệt phải → trái, nhân dồn suffix vào `answer`.
- Độ phức tạp: `O(n)`, bộ nhớ `O(1)` (ngoại trừ output).


```python
# phương pháp 1: Brute Force (dùng 2 vòng lặp O(n^2))

def product_except_self_brute_force(nums):
    n = len(nums)
    result = []
    for i in range(n):
        product = 1
        for j in range(n):
            if i != j:
                product *= nums[j]
        result.append(product)
    return result

# Time complexity: O(n^2)
# Space complexity: O(1) (không tính kết quả trả về)
nums = [1, 2, 3, 4]
print(product_except_self_brute_force(nums))  # Output: [24, 12, 8, 6]
```

```python
# phương pháp 2: Sử dụng mảng phụ left và right (O(n) time, O(n) space)
def product_except_self_prefix_suffix(nums):
    n= len(nums)
    left = [1] * n
    right = [1] * n
    result = [0] * n
    
    for i in range(1, n):
        left[i] = left[i - 1] * nums[i - 1]

    for i in range(n - 2, -1, -1):
        right[i] = right[i + 1] * nums[i + 1]

    for i in range(n):
        result[i] = left[i] * right[i]

    return result

# Time complexity: O(n)
# Space complexity: O(n)
nums = [1, 2, 3, 4]
print(product_except_self_prefix_suffix(nums))  # Output: [24, 12, 8, 6]
```

```python
# phương pháp 3: Sử dụng biến tạm thời để lưu trữ tích bên trái và bên phải (O(n) time, O(1) space)
def product_except_self_optimized(nums):
    n = len(nums)
    result = [1] * n
    
    left_product = 1
    for i in range(n):
        result[i] = left_product
        left_product *= nums[i]

    right_product = 1
    for i in range(n - 1, -1, -1):
        result[i] *= right_product
        right_product *= nums[i]

    return result
# Time complexity: O(n)
# Space complexity: O(1) (không tính kết quả trả về)
nums = [1, 2, 3, 4]
print(product_except_self_optimized(nums))  # Output: [24, 12, 8, 6]
```
